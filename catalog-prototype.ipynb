```python
import numpy as np
from catalog import draw, COUPLED, LOWPASS
```

```python
fo = 2.3e9
bw = 26.9e6
# From the Electronic Filter Design Handbook, p5-21
for name, table in list(COUPLED.items()):
    for d in table:
        qk = np.concatenate((d[:1], d[2:], d[1:2]))
        print('{} Pole {}'.format(len(qk)-1, name))
        draw(qk, bw, fo)
```

```python
import ness
fo = 2.3e9
bw = 26.9e6
# from chebyshev calculator and the low pass prototype tables in zverev
for name, table in list(LOWPASS.items()):
    for d in table:
        qk = ness.coupling_g(d)
        print('{} Pole {}'.format(len(qk)-1, name))
        draw(qk, bw, fo)
```
